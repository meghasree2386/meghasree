```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
from sklearn.metrics import mean_absolute_error, mean_squared_error
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, LSTM, Dropout
import warnings
warnings.filterwarnings('ignore')


data = pd.read_csv("final_data.csv")
data['Date'] = pd.to_datetime(data['Date'])
data = data.sort_values('Date').reset_index(drop=True)


data['Views'] = data['Views'].interpolate(method='linear')


scaler = MinMaxScaler(feature_range=(0, 1))
scaled_views = scaler.fit_transform(data[['Views']])
def create_sequences(data, seq_length):
    X, y = [], []
    for i in range(len(data) - seq_length):
        X.append(data[i:(i + seq_length)])
        y.append(data[i + seq_length])
    return np.array(X), np.array(y)


sequence_length = 30
X, y = create_sequences(scaled_views, sequence_length)


train_size = int(len(X) * 0.8)
X_train, X_test = X[:train_size], X[train_size:]
y_train, y_test = y[:train_size], y[train_size:]


model = Sequential([
    LSTM(units=50, return_sequences=True, input_shape=(X_train.shape[1], 1)),
    Dropout(0.2),
    LSTM(units=50, return_sequences=False),
    Dropout(0.2),
    Dense(units=1)
])

model.compile(optimizer='adam', loss='mean_squared_error')
history = model.fit(X_train, y_train, epochs=20, batch_size=16, validation_split=0.1, verbose=1)


predictions_scaled = model.predict(X_test)

predictions = scaler.inverse_transform(predictions_scaled)
y_test_actual = scaler.inverse_transform(y_test)


mae = mean_absolute_error(y_test_actual, predictions)
rmse = np.sqrt(mean_squared_error(y_test_actual, predictions))
print(f"\nEvaluation Metrics on Test Set:")
print(f"Mean Absolute Error (MAE): {mae:.4f}")
print(f"Root Mean Squared Error (RMSE): {rmse:.4f}")
plt.figure(figsize=(14, 6))
test_dates = data['Date'].iloc[train_size + sequence_length:].reset_index(drop=True)
plt.plot(test_dates, y_test_actual, label='Actual Traffic', color='blue', alpha=0.7)
plt.plot(test_dates, predictions, label='Predicted Traffic (LSTM)', color='orange', linestyle='--')
plt.title('Page-View Traffic Forecasting: Actual vs Predicted')
plt.xlabel('Date')
plt.ylabel('Views')
plt.legend()
plt.grid(True)
plt.show()


last_sequence = scaled_views[-sequence_length:]
future_predictions_scaled = []

current_sequence = last_sequence.reshape((1, sequence_length, 1))
for _ in range(7):
    pred = model.predict(current_sequence)
    future_predictions_scaled.append(pred[0, 0])

    current_sequence = np.append(current_sequence[:, 1:, :], pred.reshape(1, 1, 1), axis=1)

future_predictions = scaler.inverse_transform(np.array(future_predictions_scaled).reshape(-1, 1))

last_date = data['Date'].iloc[-1]
future_dates = pd.date_range(start=last_date + pd.Timedelta(days=1), periods=7)

print("\nForecasted Page Views for Unseen Future Dates:")
for date, val in zip(future_dates, future_predictions.flatten()):
    print(f"{date.strftime('%Y-%m-%d')}: {val:.2f} views")
```
